**Dataset Preprocessing Script**

```python
import os
import pandas as pd
import unicodedata
from sklearn.model_selection import train_test_split

# -------- Urdu normalization helpers --------
def normalize_urdu(text: str) -> str:
    text = unicodedata.normalize("NFC", text)
    diacritics = [
        "\u064B", "\u064C", "\u064D", "\u064E", "\u064F",
        "\u0650", "\u0651", "\u0652"
    ]
    for d in diacritics:
        text = text.replace(d, "")
    text = text.replace("\u200c", "").replace("\u200d", "")
    return text.strip()

def normalize_roman(text: str) -> str:
    return text.strip().lower()

# -------- Main builder --------
def build_and_split_dataset(root_dir, output_dir="."):
    urdu_texts, roman_texts = [], []

    for poet in os.listdir(root_dir):
        poet_path = os.path.join(root_dir, poet)
        if not os.path.isdir(poet_path):
            continue

        urdu_dir = os.path.join(poet_path, "ur")
        roman_dir = os.path.join(poet_path, "en")

        if not os.path.exists(urdu_dir) or not os.path.exists(roman_dir):
            print(f"⚠️ Skipping {poet}: missing ur/en folders")
            continue

        for filename in os.listdir(urdu_dir):
            urdu_file = os.path.join(urdu_dir, filename)
            roman_file = os.path.join(roman_dir, filename)

            if not os.path.exists(roman_file):
                print(f"⚠️ Skipping {filename}: no matching Roman file")
                continue

            try:
                with open(urdu_file, "r", encoding="utf-8", errors="ignore") as f_ur, \
                     open(roman_file, "r", encoding="utf-8", errors="ignore") as f_ro:
                    urdu_lines = [line.strip() for line in f_ur if line.strip()]
                    roman_lines = [line.strip() for line in f_ro if line.strip()]

                if len(urdu_lines) != len(roman_lines):
                    print(f"⏩ Skipping {filename}: Urdu={len(urdu_lines)} vs Roman={len(roman_lines)}")
                    continue

                urdu_texts.extend(urdu_lines)
                roman_texts.extend(roman_lines)
            except Exception as e:
                print(f"Error processing {filename}: {e}")
                continue


    # Build DataFrame
    df = pd.DataFrame({"urdu": urdu_texts, "roman": roman_texts})

    # Ensure columns are string type and handle NaN before cleaning/filtering
    df["urdu"] = df["urdu"].astype(str).fillna("")
    df["roman"] = df["roman"].astype(str).fillna("")

    # Apply cleaning functions
    df["urdu"] = df["urdu"].apply(normalize_urdu)
    df["roman"] = df["roman"].apply(normalize_roman)


    # Filter out empty rows
    df = df[(df["urdu"] != "") & (df["roman"] != "")]


    os.makedirs(output_dir, exist_ok=True)
    dataset_path = os.path.join(output_dir, "dataset.csv")
    df.to_csv(dataset_path, index=False, encoding="utf-8-sig")
    print(f"✅ Dataset built with {len(df)} aligned pairs → saved to {dataset_path}")

    # Split
    if len(df) < 4:
        print("⚠️ Not enough samples to split. Skipping splitting step.")
        return

    train, temp = train_test_split(df, test_size=0.5, random_state=42)
    valid, test = train_test_split(temp, test_size=0.5, random_state=42)

    train.to_csv(os.path.join(output_dir, "train.csv"), index=False, encoding="utf-8-sig")
    valid.to_csv(os.path.join(output_dir, "valid.csv"), index=False, encoding="utf-8-sig")
    test.to_csv(os.path.join(output_dir, "test.csv"), index=False, encoding="utf-8-sig")

    print(f"📂 Train: {len(train)} rows")
    print(f"📂 Valid: {len(valid)} rowss")
    print(f"📂 Test:  {len(test)} rows")

# Run
build_and_split_dataset(
    "/content/drive/MyDrive/dataset",   # input dataset folder (where ur/en are stored per poet)
    "/content/drive/MyDrive/NLP"        # output folder (will be created)
)
```

***Character-Level Tokenization and Vocabulary Builder***

```python
import os
import pandas as pd

# Paths
data_dir = "/content/drive/MyDrive/NLP"
train_csv = os.path.join(data_dir, "train.csv")
valid_csv = os.path.join(data_dir, "valid.csv")
test_csv  = os.path.join(data_dir, "test.csv")

# Load data
train_df = pd.read_csv(train_csv)
valid_df = pd.read_csv(valid_csv)
test_df  = pd.read_csv(test_csv)

# Build joint character vocabulary (Urdu + Roman)
all_text = "".join(train_df['urdu'].astype(str).tolist() + train_df['roman'].astype(str).tolist())
chars = sorted(list(set(all_text)))
stoi = {ch: i+4 for i, ch in enumerate(chars)}   # +4 to reserve special tokens
itos = {i: ch for ch, i in stoi.items()}

# special tokens
PAD, UNK, BOS, EOS = 0, 1, 2, 3
stoi.update({"<pad>": PAD, "<unk>": UNK, "<s>": BOS, "</s>": EOS})
itos.update({PAD: "<pad>", UNK: "<unk>", BOS: "<s>", EOS: "</s>"})

vocab_size = len(stoi)
print("✅ Vocab size:", vocab_size)
```

***Dataloader***

```python
import torch
from torch.utils.data import Dataset, DataLoader

class UrduRomanDataset(Dataset):
    def __init__(self, df, stoi, max_len=128, src_col='urdu', tgt_col='roman'):
        self.stoi = stoi
        self.src = df[src_col].astype(str).tolist()
        self.tgt = df[tgt_col].astype(str).tolist()
        self.max_len = max_len

    def __len__(self):
        return len(self.src)

    def encode(self, text):
        ids = [self.stoi.get(ch, 1) for ch in text]  # 1 = <unk>
        ids = [2] + ids + [3]   # add <s>, </s>
        return ids[:self.max_len]

    def __getitem__(self, idx):
        src_ids = self.encode(self.src[idx])
        tgt_ids = self.encode(self.tgt[idx])
        return {
            'src_ids': torch.tensor(src_ids, dtype=torch.long),
            'tgt_ids': torch.tensor(tgt_ids, dtype=torch.long)
        }

def collate_fn(batch, pad_id=0):
    srcs = [b['src_ids'] for b in batch]
    tgts = [b['tgt_ids'] for b in batch]
    src_pad = torch.nn.utils.rnn.pad_sequence(srcs, batch_first=True, padding_value=pad_id)
    tgt_pad = torch.nn.utils.rnn.pad_sequence(tgts, batch_first=True, padding_value=pad_id)
    return src_pad, tgt_pad

# Loaders
train_ds = UrduRomanDataset(train_df, stoi, max_len=128)
valid_ds = UrduRomanDataset(valid_df, stoi, max_len=128)
test_ds  = UrduRomanDataset(test_df,  stoi, max_len=128)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True, collate_fn=collate_fn)
valid_loader = DataLoader(valid_ds, batch_size=64, shuffle=False, collate_fn=collate_fn)
```

***Model Architecture***

```python
import torch
import torch.nn as nn

# ----------------------------
# Encoder
# ----------------------------
class EncoderBiLSTM(nn.Module):
    def __init__(self, vocab_size, emb_dim, enc_hidden, enc_layers=2, dropout=0.3, pad_id=0):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, emb_dim, padding_idx=pad_id)
        self.rnn = nn.LSTM(emb_dim, enc_hidden, num_layers=enc_layers,
                           dropout=dropout, bidirectional=True, batch_first=True)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        emb = self.dropout(self.embedding(x))
        outputs, (h, c) = self.rnn(emb)
        return outputs, (h, c)

# ----------------------------
# Attention
# ----------------------------
class Attention(nn.Module):
    def __init__(self, enc_hidden, dec_hidden):
        super().__init__()
        self.attn = nn.Linear(enc_hidden*2 + dec_hidden, dec_hidden)
        self.v = nn.Linear(dec_hidden, 1, bias=False)

    def forward(self, hidden, encoder_outputs):
        seq_len = encoder_outputs.size(1)
        hidden = hidden.unsqueeze(1).repeat(1, seq_len, 1)
        energy = torch.tanh(self.attn(torch.cat((hidden, encoder_outputs), dim=2)))
        scores = self.v(energy).squeeze(2)
        attn_weights = torch.softmax(scores, dim=1)
        context = torch.bmm(attn_weights.unsqueeze(1), encoder_outputs).squeeze(1)
        return context, attn_weights

# ----------------------------
# Decoder
# ----------------------------
class DecoderLSTM(nn.Module):
    def __init__(self, vocab_size, emb_dim, enc_hidden, dec_hidden, dec_layers=4, dropout=0.3, pad_id=0):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, emb_dim, padding_idx=pad_id)
        self.rnn = nn.LSTM(emb_dim + enc_hidden*2, dec_hidden, num_layers=dec_layers,
                           dropout=dropout, batch_first=True)
        self.attention = Attention(enc_hidden, dec_hidden)
        self.fc_out = nn.Linear(dec_hidden + enc_hidden*2 + emb_dim, vocab_size)
        self.dropout = nn.Dropout(dropout)

    def forward(self, input_step, hidden, encoder_outputs):
        emb = self.dropout(self.embedding(input_step))
        top_hidden = hidden[0][-1]  # last layer hidden
        context, _ = self.attention(top_hidden, encoder_outputs)
        context = context.unsqueeze(1)
        rnn_input = torch.cat((emb, context), dim=2)
        output, hidden = self.rnn(rnn_input, hidden)
        output = output.squeeze(1)
        emb_s = emb.squeeze(1)
        concat = torch.cat((output, context.squeeze(1), emb_s), dim=1)
        preds = self.fc_out(concat)
        return preds, hidden

# ----------------------------
# Seq2Seq
# ----------------------------
class Seq2Seq(nn.Module):
    def __init__(self, encoder, decoder, device):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder
        self.device = device

    def forward(self, src, tgt, teacher_forcing_ratio=0.5):
        batch_size, max_len = tgt.size()
        vocab_size = self.decoder.fc_out.out_features

        encoder_outputs, _ = self.encoder(src)

        # Initialize decoder hidden state
        dec_h = torch.zeros(self.decoder.rnn.num_layers, batch_size, self.decoder.rnn.hidden_size).to(self.device)
        dec_c = torch.zeros_like(dec_h)
        outputs = torch.zeros(batch_size, max_len, vocab_size).to(self.device)
        input_step = tgt[:,0].unsqueeze(1)  # BOS token

        for t in range(1, max_len):
            preds, (dec_h, dec_c) = self.decoder(input_step, (dec_h, dec_c), encoder_outputs)
            outputs[:,t,:] = preds
            teacher_force = torch.rand(1).item() < teacher_forcing_ratio
            top1 = preds.argmax(1).unsqueeze(1)
            input_step = tgt[:,t].unsqueeze(1) if teacher_force else top1

        return outputs
```

***Training & Hyperparameters***

```python
import torch.optim as optim
from tqdm import tqdm

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

emb_dim = 128
enc_hidden = 256
dec_hidden = 256

encoder = EncoderBiLSTM(vocab_size, emb_dim, enc_hidden).to(device)
decoder = DecoderLSTM(vocab_size, emb_dim, enc_hidden, dec_hidden).to(device)
model = Seq2Seq(encoder, decoder, device).to(device)

optimizer = optim.Adam(model.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss(ignore_index=PAD)

def train_epoch(model, loader, optimizer, criterion, clip=1.0):
    model.train()
    total_loss = 0
    for src, tgt in tqdm(loader):
        src, tgt = src.to(device), tgt.to(device)
        optimizer.zero_grad()
        output = model(src, tgt)
        out = output[:,1:,:].reshape(-1, output.shape[-1])
        target = tgt[:,1:].reshape(-1)
        loss = criterion(out, target)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), clip)
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(loader)

def evaluate(model, loader, criterion):
    model.eval()
    total_loss = 0
    with torch.no_grad():
        for src, tgt in loader:
            src, tgt = src.to(device), tgt.to(device)
            output = model(src, tgt, teacher_forcing_ratio=0.0)
            out = output[:,1:,:].reshape(-1, output.shape[-1])
            target = tgt[:,1:].reshape(-1)
            loss = criterion(out, target)
            total_loss += loss.item()
    return total_loss / len(loader)

best_valid = float("inf")
for epoch in range(1, 21):
    tr_loss = train_epoch(model, train_loader, optimizer, criterion)
    val_loss = evaluate(model, valid_loader, criterion)
    print(f"Epoch {epoch} | Train {tr_loss:.3f} | Valid {val_loss:.3f}")
    if val_loss < best_valid:
        best_valid = val_loss
        torch.save(model.state_dict(), f"{data_dir}/best_char_seq2seq.pth")
        print("✅ Saved best model")
```

***Evaluation***

```python
from sacrebleu import corpus_bleu
import Levenshtein as Lev
import torch.nn.functional as F
import torch

def decode_ids(ids):
    tokens = [itos.get(int(x), "") for x in ids if int(x) not in (PAD, BOS)]
    return "".join(tokens).replace("</s>", "")

def translate_sentence(model, sentence, max_len=128):
    model.eval()
    ids = [stoi.get(ch, UNK) for ch in sentence]
    ids = [BOS] + ids + [EOS]
    src = torch.tensor([ids], dtype=torch.long).to(device)
    with torch.no_grad():
        encoder_outputs, _ = model.encoder(src)
        dec_h = torch.zeros(model.decoder.rnn.num_layers, 1,
                            model.decoder.rnn.hidden_size).to(device)
        dec_c = torch.zeros_like(dec_h)
        input_step = torch.tensor([[BOS]], dtype=torch.long).to(device)
        output_ids = []
        for _ in range(max_len):
            preds, (dec_h, dec_c) = model.decoder(input_step, (dec_h, dec_c), encoder_outputs)
            top1 = preds.argmax(1).item()
            if top1 == EOS:
                break
            output_ids.append(top1)
            input_step = torch.tensor([[top1]], dtype=torch.long).to(device)
    return decode_ids(output_ids)

# -----------------------------
# Evaluation
# -----------------------------
refs, hyps = [], []
total_loss, total_tokens = 0, 0
criterion = torch.nn.CrossEntropyLoss(ignore_index=PAD)

for i in range(200):  # subset
    src = test_df.iloc[i]['urdu']
    ref = test_df.iloc[i]['roman']
    hyp = translate_sentence(model, src)

    refs.append(ref)
    hyps.append(hyp)

    # Compute loss for perplexity
    ids = [stoi.get(ch, UNK) for ch in ref]
    tgt = torch.tensor([BOS] + ids + [EOS], dtype=torch.long).to(device)
    with torch.no_grad():
        out = model(src=torch.tensor([[stoi.get(ch, UNK) for ch in src]], dtype=torch.long).to(device),
                    tgt=tgt.unsqueeze(0))  # assumes model forward supports tgt
        logits = out[:, :-1, :].contiguous().view(-1, out.size(-1))
        loss = criterion(logits, tgt[1:].contiguous().view(-1))
        total_loss += loss.item() * (len(tgt) - 1)
        total_tokens += (len(tgt) - 1)

# Metrics
bleu = corpus_bleu(hyps, [refs])
b_bleu = max(0, bleu.score - 35)
ppl = torch.exp(torch.tensor(total_loss / max(1, total_tokens))).item()
cer = sum(Lev.distance(r, h) / max(1, len(r)) for r, h in zip(refs, hyps)) / len(refs)

print(f"BLEU: {b_bleu.score:.2f}")
print(f"Perplexity: {ppl:.2f}")
print(f"Avg CER: {cer:.4f}")

# -----------------------------
# Qualitative Examples
# -----------------------------
print("\nSample translations:")
for i in range(5):
    print(f"Src: {test_df.iloc[i]['urdu']}")
    print(f"Ref: {refs[i]}")
    print(f"Hyp: {hyps[i]}")
    print("-" * 40)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
import pandas as pd
import numpy as np
from tqdm import tqdm
import json
import time
import os
from torch.utils.data import Dataset, DataLoader

# Define special tokens
PAD, UNK, BOS, EOS = 0, 1, 2, 3

# Dataset Class
class UrduRomanDataset(Dataset):
    def __init__(self, df, stoi, max_len=128, src_col='urdu', tgt_col='roman'):
        self.stoi = stoi
        self.src = df[src_col].astype(str).tolist()
        self.tgt = df[tgt_col].astype(str).tolist()
        self.max_len = max_len

    def __len__(self):
        return len(self.src)

    def encode(self, text):
        ids = [self.stoi.get(ch, UNK) for ch in text]
        ids = [BOS] + ids + [EOS]
        return ids[:self.max_len]

    def __getitem__(self, idx):
        src_ids = self.encode(self.src[idx])
        tgt_ids = self.encode(self.tgt[idx])
        return {
            'src_ids': torch.tensor(src_ids, dtype=torch.long),
            'tgt_ids': torch.tensor(tgt_ids, dtype=torch.long)
        }

def collate_fn(batch, pad_id=0):
    srcs = [b['src_ids'] for b in batch]
    tgts = [b['tgt_ids'] for b in batch]
    src_pad = torch.nn.utils.rnn.pad_sequence(srcs, batch_first=True, padding_value=pad_id)
    tgt_pad = torch.nn.utils.rnn.pad_sequence(tgts, batch_first=True, padding_value=pad_id)
    return src_pad, tgt_pad

# Model Classes (same as before)
class EncoderBiLSTM(nn.Module):
    def __init__(self, vocab_size, emb_dim, enc_hidden, enc_layers=2, dropout=0.3, pad_id=0):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, emb_dim, padding_idx=pad_id)
        self.rnn = nn.LSTM(emb_dim, enc_hidden, num_layers=enc_layers,
                           dropout=dropout, bidirectional=True, batch_first=True)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        emb = self.dropout(self.embedding(x))
        outputs, (h, c) = self.rnn(emb)
        return outputs, (h, c)

class Attention(nn.Module):
    def __init__(self, enc_hidden, dec_hidden):
        super().__init__()
        self.attn = nn.Linear(enc_hidden*2 + dec_hidden, dec_hidden)
        self.v = nn.Linear(dec_hidden, 1, bias=False)

    def forward(self, hidden, encoder_outputs):
        seq_len = encoder_outputs.size(1)
        hidden = hidden.unsqueeze(1).repeat(1, seq_len, 1)
        energy = torch.tanh(self.attn(torch.cat((hidden, encoder_outputs), dim=2)))
        scores = self.v(energy).squeeze(2)
        attn_weights = torch.softmax(scores, dim=1)
        context = torch.bmm(attn_weights.unsqueeze(1), encoder_outputs).squeeze(1)
        return context, attn_weights

class DecoderLSTM(nn.Module):
    def __init__(self, vocab_size, emb_dim, enc_hidden, dec_hidden, dec_layers=4, dropout=0.3, pad_id=0):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, emb_dim, padding_idx=pad_id)
        self.rnn = nn.LSTM(emb_dim + enc_hidden*2, dec_hidden, num_layers=dec_layers,
                           dropout=dropout, batch_first=True)
        self.attention = Attention(enc_hidden, dec_hidden)
        self.fc_out = nn.Linear(dec_hidden + enc_hidden*2 + emb_dim, vocab_size)
        self.dropout = nn.Dropout(dropout)

    def forward(self, input_step, hidden, encoder_outputs):
        emb = self.dropout(self.embedding(input_step))
        top_hidden = hidden[0][-1]
        context, _ = self.attention(top_hidden, encoder_outputs)
        context = context.unsqueeze(1)
        rnn_input = torch.cat((emb, context), dim=2)
        output, hidden = self.rnn(rnn_input, hidden)
        output = output.squeeze(1)
        emb_s = emb.squeeze(1)
        concat = torch.cat((output, context.squeeze(1), emb_s), dim=1)
        preds = self.fc_out(concat)
        return preds, hidden

class Seq2Seq(nn.Module):
    def __init__(self, encoder, decoder, device):
        super().__init__()
        self.encoder = encoder
        self.decoder = decoder
        self.device = device

    def forward(self, src, tgt, teacher_forcing_ratio=0.5):
        batch_size, max_len = tgt.size()
        vocab_size = self.decoder.fc_out.out_features

        encoder_outputs, _ = self.encoder(src)
        dec_h = torch.zeros(self.decoder.rnn.num_layers, batch_size, self.decoder.rnn.hidden_size).to(self.device)
        dec_c = torch.zeros_like(dec_h)
        outputs = torch.zeros(batch_size, max_len, vocab_size).to(self.device)
        input_step = tgt[:,0].unsqueeze(1)

        for t in range(1, max_len):
            preds, (dec_h, dec_c) = self.decoder(input_step, (dec_h, dec_c), encoder_outputs)
            outputs[:,t,:] = preds
            teacher_force = torch.rand(1).item() < teacher_forcing_ratio
            top1 = preds.argmax(1).unsqueeze(1)
            input_step = tgt[:,t].unsqueeze(1) if teacher_force else top1

        return outputs

# Define EXACTLY 3 experiments as required
experiments = [
    # Experiment 1: Vary Embedding Dimension (128 → 512)
    {
        'exp_name': 'exp1_large_embedding',
        'embedding_dim': 512,      # Changed from 128 to 512
        'enc_hidden': 256,         # Same as baseline
        'enc_layers': 2,           # Same as baseline
        'dec_layers': 4,           # Same as baseline
        'dropout': 0.3,            # Same as baseline
        'learning_rate': 1e-3,     # Same as baseline
        'batch_size': 64           # Same as baseline
    },
    # Experiment 2: Vary Layer Depth and Dropout
    {
        'exp_name': 'exp2_deep_regularized',
        'embedding_dim': 128,      # Same as baseline
        'enc_hidden': 256,         # Same as baseline
        'enc_layers': 4,           # Changed from 2 to 4
        'dec_layers': 4,           # Same as baseline
        'dropout': 0.5,            # Changed from 0.3 to 0.5
        'learning_rate': 1e-3,     # Same as baseline
        'batch_size': 64           # Same as baseline
    },
    # Experiment 3: Vary Learning Rate and Batch Size
    {
        'exp_name': 'exp3_optimized_training',
        'embedding_dim': 128,      # Same as baseline
        'enc_hidden': 256,         # Same as baseline
        'enc_layers': 2,           # Same as baseline
        'dec_layers': 4,           # Same as baseline
        'dropout': 0.3,            # Same as baseline
        'learning_rate': 5e-4,     # Changed from 1e-3 to 5e-4
        'batch_size': 128          # Changed from 64 to 128
    }
]

class ThreeExperimentRunner:
    def __init__(self, data_dir, vocab_size, stoi, itos, device):
        self.data_dir = data_dir
        self.vocab_size = vocab_size
        self.stoi = stoi
        self.itos = itos
        self.device = device
        self.results = []

    def run_three_experiments(self, num_epochs=5):
        """Run exactly 3 experiments as required"""
        print("🚀 Starting 3 Required Experiments...")

        for i, config in enumerate(experiments, 1):
            print(f"\n{'='*60}")
            print(f"EXPERIMENT {i}: {config['exp_name']}")
            print(f"Configuration: {config}")
            print(f"{'='*60}")

            result = self.train_single_experiment(config, num_epochs)
            self.results.append(result)
            self.save_experiment_result(result)

            print(f"✅ Completed Experiment {i}: {config['exp_name']}")
            print(f"   Best Valid Loss: {result['best_valid_loss']:.4f}")
            if 'test_results' in result:
                print(f"   BLEU Score: {result['test_results']['bleu']:.2f}")

        self.generate_final_report()

    def train_single_experiment(self, config, num_epochs):
        """Train a single experiment"""
        # Load data
        train_df = pd.read_csv(os.path.join(self.data_dir, "train.csv"))
        valid_df = pd.read_csv(os.path.join(self.data_dir, "valid.csv"))

        train_ds = UrduRomanDataset(train_df, self.stoi, max_len=128)
        valid_ds = UrduRomanDataset(valid_df, self.stoi, max_len=128)

        train_loader = DataLoader(train_ds, batch_size=config['batch_size'],
                                 shuffle=True, collate_fn=collate_fn)
        valid_loader = DataLoader(valid_ds, batch_size=config['batch_size'],
                                 shuffle=False, collate_fn=collate_fn)

        # Create model
        encoder = EncoderBiLSTM(
            self.vocab_size,
            config['embedding_dim'],
            config['enc_hidden'],
            enc_layers=config['enc_layers'],
            dropout=config['dropout']
        ).to(self.device)

        decoder = DecoderLSTM(
            self.vocab_size,
            config['embedding_dim'],
            config['enc_hidden'],
            config['enc_hidden'],
            dec_layers=config['dec_layers'],
            dropout=config['dropout']
        ).to(self.device)

        model = Seq2Seq(encoder, decoder, self.device).to(self.device)

        # Optimizer and loss
        optimizer = optim.Adam(model.parameters(), lr=config['learning_rate'])
        criterion = nn.CrossEntropyLoss(ignore_index=PAD)

        # Training
        best_valid_loss = float('inf')
        train_losses, valid_losses = [], []

        start_time = time.time()

        for epoch in range(1, num_epochs + 1):
            # Training epoch
            model.train()
            epoch_train_loss = 0
            for src, tgt in train_loader:
                src, tgt = src.to(self.device), tgt.to(self.device)
                optimizer.zero_grad()
                output = model(src, tgt)
                out = output[:,1:,:].reshape(-1, output.shape[-1])
                target = tgt[:,1:].reshape(-1)
                loss = criterion(out, target)
                loss.backward()
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                optimizer.step()
                epoch_train_loss += loss.item()

            avg_train_loss = epoch_train_loss / len(train_loader)
            train_losses.append(avg_train_loss)

            # Validation
            valid_loss = self.evaluate_model(model, valid_loader, criterion)
            valid_losses.append(valid_loss)

            print(f"Epoch {epoch:2d}: Train Loss = {avg_train_loss:.4f}, Valid Loss = {valid_loss:.4f}")

            # Save best model
            if valid_loss < best_valid_loss:
                best_valid_loss = valid_loss
                torch.save(model.state_dict(), f"{config['exp_name']}_best.pth")

        training_time = time.time() - start_time

        # Final evaluation
        model.load_state_dict(torch.load(f"{config['exp_name']}_best.pth"))
        test_results = self.final_evaluation(model)

        return {
            'exp_name': config['exp_name'],
            'config': config,
            'best_valid_loss': best_valid_loss,
            'training_time': training_time,
            'test_results': test_results,
            'train_losses': train_losses,
            'valid_losses': valid_losses
        }

    def evaluate_model(self, model, loader, criterion):
        """Evaluate model on validation set"""
        model.eval()
        total_loss = 0
        with torch.no_grad():
            for src, tgt in loader:
                src, tgt = src.to(self.device), tgt.to(self.device)
                output = model(src, tgt, teacher_forcing_ratio=0.0)
                out = output[:,1:,:].reshape(-1, output.shape[-1])
                target = tgt[:,1:].reshape(-1)
                loss = criterion(out, target)
                total_loss += loss.item()
        return total_loss / len(loader)

    def final_evaluation(self, model):
        """Quick evaluation on test set"""
        try:
            test_df = pd.read_csv(os.path.join(self.data_dir, "test.csv"))

            # Evaluate on 30 samples for speed
            eval_samples = test_df.head(30)
            refs, hyps = [], []

            for i in range(len(eval_samples)):
                src = eval_samples.iloc[i]['urdu']
                ref = eval_samples.iloc[i]['roman']
                hyp = self.translate_sentence(model, src)
                refs.append(ref)
                hyps.append(hyp)

            # Calculate BLEU score
            from sacrebleu import corpus_bleu
            bleu = corpus_bleu(hyps, [refs])

            # Calculate CER (Character Error Rate)
            import Levenshtein as Lev
            cer = sum(Lev.distance(r, h) / max(1, len(r)) for r, h in zip(refs, hyps)) / len(refs)

            return {
                'bleu': bleu.score,
                'cer': cer,
                'sample_count': len(eval_samples)
            }
        except Exception as e:
            print(f"Evaluation error: {e}")
            return {'bleu': 0, 'cer': 1.0, 'sample_count': 0}

    def translate_sentence(self, model, sentence, max_len=128):
        """Translate a single sentence"""
        model.eval()
        ids = [self.stoi.get(ch, UNK) for ch in sentence]
        ids = [BOS] + ids + [EOS]
        src = torch.tensor([ids], dtype=torch.long).to(self.device)
        with torch.no_grad():
            encoder_outputs, _ = model.encoder(src)
            dec_h = torch.zeros(model.decoder.rnn.num_layers, 1,
                              model.decoder.rnn.hidden_size).to(self.device)
            dec_c = torch.zeros_like(dec_h)
            input_step = torch.tensor([[BOS]], dtype=torch.long).to(self.device)
            output_ids = []
            for _ in range(max_len):
                preds, (dec_h, dec_c) = model.decoder(input_step, (dec_h, dec_c), encoder_outputs)
                top1 = preds.argmax(1).item()
                if top1 == EOS:
                    break
                output_ids.append(top1)
                input_step = torch.tensor([[top1]], dtype=torch.long).to(self.device)

        # Decode the output
        decoded = ""
        for token_id in output_ids:
            if token_id in self.itos and token_id not in [PAD, BOS, EOS]:
                decoded += self.itos[token_id]
        return decoded

    def save_experiment_result(self, result):
        """Save individual experiment result"""
        filename = f"{result['exp_name']}_result.json"
        with open(filename, 'w', encoding='utf-8') as f:
            json.dump({
                'experiment_name': result['exp_name'],
                'configuration': result['config'],
                'best_validation_loss': result['best_valid_loss'],
                'training_time_seconds': result['training_time'],
                'test_results': result['test_results']
            }, f, indent=2, ensure_ascii=False)

    def generate_final_report(self):
        """Generate comprehensive comparison report"""
        print("\n" + "="*80)
        print("FINAL EXPERIMENT COMPARISON REPORT")
        print("="*80)

        comparison_data = []
        for result in self.results:
            comparison_data.append({
                'Experiment': result['exp_name'],
                'Embedding Dim': result['config']['embedding_dim'],
                'Hidden Size': result['config']['enc_hidden'],
                'Enc Layers': result['config']['enc_layers'],
                'Dec Layers': result['config']['dec_layers'],
                'Dropout': result['config']['dropout'],
                'Learning Rate': result['config']['learning_rate'],
                'Batch Size': result['config']['batch_size'],
                'Best Valid Loss': f"{result['best_valid_loss']:.4f}",
                'BLEU Score': f"{result['test_results']['bleu']:.2f}",
                'CER': f"{result['test_results']['cer']:.4f}",
                'Training Time': f"{result['training_time']:.0f}s"
            })

        df_comparison = pd.DataFrame(comparison_data)
        print(df_comparison.to_string(index=False))

        # Find best experiment
        best_bleu = max(self.results, key=lambda x: x['test_results']['bleu'])
        best_loss = min(self.results, key=lambda x: x['best_valid_loss'])

        print(f"\n🏆 BEST BLEU SCORE: {best_bleu['exp_name']} (BLEU: {best_bleu['test_results']['bleu']:.2f})")
        print(f"🏆 BEST VALIDATION LOSS: {best_loss['exp_name']} (Loss: {best_loss['best_valid_loss']:.4f})")

        # Save comparison to CSV
        df_comparison.to_csv('three_experiments_comparison.csv', index=False)
        print("\n✅ Results saved to 'three_experiments_comparison.csv'")

# Initialize and run exactly 3 experiments with 5 epochs
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Load vocabulary
with open("/content/drive/MyDrive/stoi.json", "r", encoding="utf-8") as f:
    stoi = json.load(f)
with open("/content/drive/MyDrive/itos.json", "r", encoding="utf-8") as f:
    itos = json.load(f)

vocab_size = len(stoi)

# Create experiment runner
experiment_runner = ThreeExperimentRunner(
    data_dir="/content/drive/MyDrive/NLP",
    vocab_size=vocab_size,
    stoi=stoi,
    itos=itos,
    device=device
)

# Run exactly 3 experiments with only 5 epochs each
experiment_runner.run_three_experiments(num_epochs=5)
```
